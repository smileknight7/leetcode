```python
# 度--某个顶点与其他顶点直接相连的edge数， 对于有向图，度分为入度和出度


# 联通图与联通分量：连通图是指，图中任意两个顶点之间都存在一条路径，如果图可以划分为多个子图，
# 使得每个子图都是连通的，那么这些子图就是联通分量
# 在有向图中，如果对于任意两个顶点a和b，从a到b和从b到a都存在路径，
# 则称其为强连通图，有向图中的极大强连通子图被成为这个有向图的强联通分量（全连接）

# 生成树是全连接图的一个子图，这颗树包含了图中所有顶点，生成树可用于图遍，最小生成树等问题。



# 邻接矩阵: 对于有n个顶点的图，邻接矩阵是n x n 的方阵， 对于邻接矩阵不需要考虑是有向还是无向的， 有向图可以兼容无向图
# 邻接矩阵不适用于稀疏图

# 邻接矩阵优点是代码简单实现，但是缺点是占用空间很大



# 邻接表：适用于稀疏图
# 邻接表有两种存储方式，顺序存储和链式存储
# 邻接表的应用: DFS, BFS, 最小生成树, 最短路径等问题

# 邻接表优点：节省空间，读取速度快
# 邻接表缺点：不适合存储稠密图，每个顶点的边列表过长降低存储和访问效率

```

```python


# 邻接矩阵
inf = -1
class Graph:
    def __init__(self, n):
        self.n = n
        self.edges = []
        for i in range(n):                      # 构建nXn矩阵
            l = []
            for j in range(n):
                l.append(inf)
            self.edges.append(l)

    def addEdge(self, u, v , w):
        self.edges[u][v] = w


    def printGraph(self):
        for i in range(self.n):
            for j in range(self.n):
                print(self.edges[i][j], end = " ")
            print()
def Test():
    graph = Graph(5)
    graph.addEdge(0, 1, 5)
    graph.addEdge(0, 2, 5)
    graph.addEdge(1, 2, 5)
    graph.addEdge(2, 0, 5)
    graph.addEdge(2, 3, 5)
    graph.addEdge(3, 3, 5)
    graph.printGraph()
Test()




# 邻接表模板

class EdgeNode:
    def __init__(self, v, w):
        self.vertex = v    # 弧尾
        self.weight = w    # 权重

class VertexNode:
    def __init__(self, v):
        self.vertex = v    # 顶点
        self.edges = []    # 边表


class Graph2:
    def __init__(self, n):
        self.n = n 
        self.nodes = [VertexNode(i) for i in range(n)]  # 顶点表(顺序表)


    def addEdge(self, u, v, w):
        edge = EdgeNode(v, w)
        self.nodes[u].edges.append(edge)            # 通过弧头来索引到edge表上然后addedge


    def printGraph(self):
        for i in range(self.n):
            print('vertex', i , end = " : ")
            for edge in self.nodes[i].edges:
                print('->', edge.vertex, 'weight:', edge.weight, end = " ")
            print()

def Test():
    graph = Graph2(5)
    graph.addEdge(0, 1, 5)
    graph.addEdge(0, 2, 5)
    graph.addEdge(1, 2, 5)
    graph.addEdge(2, 0, 5)
    graph.addEdge(2, 3, 5)
    graph.addEdge(3, 3, 5)
    graph.printGraph()

Test()
```
