

# tree的定义

## 树是n个节点的有限集合，树的概念中涉及到递归的问题

# ！！！ 只要找到环在树中，就证明它不是一个树状



```python


# tree


class TreeNode:
    def __init__(self):
        self.data = None
        self.children = []
    
    def AddChild(self, child):
        self.children.append(child)
        
class Tree:
    def __init__(self, maxNodes):
        self.root = None
        self.nodes = [TreeNode() for _ in range(maxNodes)]           # 这个是个treenode的缓存池
    
    def getTreeNode(self, id):
        return self.nodes[id]
    
    def setRoot(self, rootId):
        self.root = self.getTreeNode(rootId)

    def addChild(self, parentId, childId):                          # 构建边的部分
        parentNode = self.getTreeNode(parentId)
        childNode = self.getTreeNode(childId)
        parentNode.AddChild(childNode)

    def AssignData(self, id, data):
        self.getTreeNode(id).data = data

    def Print(self, node=None):                          # 深度优先遍历（这里那里体现了回溯呢？）
        if node == None:
            node = self.root
        print(node.data)
        for child in node.children:
            self.Print(child)                       # 递归打印树的节点数据(对一个节点一直走到头才会返回头走下个节点)


def Test():
    T = Tree(10)
    T.setRoot(0)
    T.AssignData(0, "root")
    T.AssignData(1, "A")                            # 构建节点
    T.AssignData(2, "B")
    T.AssignData(3, "C")
    T.AssignData(4, "D")
    
    T.addChild(0, 1)                                # 构建edge
    T.addChild(0, 2)
    T.addChild(1, 3)
    T.Print()
Test()
    
```

```python


# 二叉树三个节点组成，判断根节点是否等于两个子节点之和


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right



def checkTree(root):
    root = TreeNode(10, TreeNode(4), TreeNode(6))
    return root.val == root.left.val + root.right.val



print(checkTree(TreeNode(10, TreeNode(4), TreeNode(6))))  # 输出 True


# 二叉树最大深度（根节点到最远节点的节点数量）


# → maxDepth(3)
#     → maxDepth(9)
#         maxDepth(None) = 0
#         maxDepth(None) = 0
#     ← maxDepth(9) = max(0, 0) + 1 = 1
#     → maxDepth(20)
#         → maxDepth(15)
#             maxDepth(None) = 0
#             maxDepth(None) = 0
#         ← maxDepth(15) = max(0, 0) + 1 = 1
#         → maxDepth(7)
#             maxDepth(None) = 0
#             maxDepth(None) = 0
#         ← maxDepth(7) = max(0, 0) + 1 = 1
#     ← maxDepth(20) = max(1, 1) + 1 = 2
# ← maxDepth(3) = max(1, 2) + 1 = 3



def maxDepth(root):                 # 这里使用递归的方法
    
    if root == None:
        return 0
    ld = maxDepth(root.left)
    rd = maxDepth(root.right)
    return max(ld, rd) + 1

root = TreeNode(3, TreeNode(9), TreeNode(20, TreeNode(15), TreeNode(7)))
print(maxDepth(root))



# 计算二叉深度
class Solution:

    def __init__(self):
        pass
    
    def dfs(self, root, depth):
        if depth > self.maxDepth:
            self.maxDepth = depth
        if root:
            self.dfs(root.left, depth + 1)
            self.dfs(root.right, depth + 1)

    def calculateDepth(self, root):
        self.maxDepth = 0
        self.dfs(root, 0)
        return self.maxDepth

```

# 二叉树

## 二叉树的遍历方式有，前序遍历，中序遍历，后序遍历，层序遍历


## 前序遍历
### 根节点---->左节点--->右节点

## 中序遍历
### 左节点---->根节点--->右节点

## 后序遍历
### 左节点---->右节点--->根节点

## 层序遍历
### BFS 广度优先搜索

```python


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Tree:
    def __init__(self, maxNodes):
        self.root = None
        self.nodes = [TreeNode() for _ in range(maxNodes)]
        self.nodeSize = maxNodes
    
    def GetTreeNode(self, id):
        return self.nodes[id]
    
    def visit(self, node):
        print(node.val, end=" ")
    
    def Create(self, a, size, nodeId):              # 这里是构建了一个完全二叉树
        if nodeId >= size or a[nodeId] == None:
            return None
        nowNode = self.GetTreeNode(nodeId)
        nowNode.val = a[nodeId]
        nowNode.left = self.Create(a, size, nodeId * 2)
        nowNode.right = self.Create(a, size, nodeId * 2 + 1)
        return nowNode
    
    def CreateTree(self, a):
        self.root = self.Create(a,len(a), 1)
```

## 前序遍历  先访问root然后左边子树，然后再访问右子树(也就是说左子树会一直访问到底部的)

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def preorder(root, ret):
    if root == None:
        return
    ret.append(root.val)                         # 访问根节点
    preorder(root.left, ret)            # 访问左子树
    preorder(root.right, ret)

def predorderTraversal(root):
    ret = []
    preorder(root, ret)
    return ret

root = TreeNode(1)
root.right = TreeNode(2)
root.right.left = TreeNode(3)
print(predorderTraversal(root))  # 输出 [1, 2, 3]


```

## 中序遍历     先访问左子树， 再访问根节点，最后访问右子树 （一开始是从左边最端点开始访问）

```python

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def inorder(root, ret):
    if root == None:
        return
    inorder(root.left, ret)            # 访问左子树
    ret.append(root.val)                         # 访问根节点
    inorder(root.right, ret)           # 访问右子树

def inorderTraversal(root):
    ret = []
    inorder(root, ret)
    return ret

root = TreeNode(1)
root.right = TreeNode(2)
root.right.left = TreeNode(3)
print(inorderTraversal(root))  # 输出 [1, 3, 2]

```

## 后序遍历    先访问左子树，再访问右边子树状，最后再访问根节点

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def postorder(root, ret):
    if root == None:
        return
    postorder(root.left, ret)            # 访问左子树
    postorder(root.right, ret)           # 访问右子树
    ret.append(root.val)                         # 访问根节点

def postorderTraversal(root):
    ret = []
    postorder(root, ret)
    return ret

root = TreeNode(1)
root.right = TreeNode(2)
root.right.left = TreeNode(3)
print(postorderTraversal(root))  # 输出 [3, 2, 1]
```

## 层序遍历   广度优先搜索(使用对列来解决)

## 二叉搜索树  ----- 中序遍历一定是一个有序列表（因为左边小于根节点小于右边）
### 空树是二叉搜索树状
### 若左子树不为空，则左子树上所有节点的值均小于其根节点的值
### 若右子树不为空，则右子树上所有节点值均大于其根节点值
### 其左右子树都是二叉搜索树




```python
class Treenode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insertNode(self, root, val):
        if root is None:
            return Treenode(val)                                # 插入一定要在叶子节点
        
        if val < root.val:
            root.left = self.insertNode(root.left, val)    # 递归插入到左子树
        elif val > root.val:
            root.right = self.insertNode(root.right, val)    # 插入到右子树
        return root
    
    def insert(self, val):
        self.root = self.insertNode(self.root, val)
    

    def removeNode(self, root, val):
        if root is None:
            return None
        
        if val < root.val:                                          # 更新过的子树接在父节点上
            root.left = self.removeNode(root.left, val)
        elif val > root.val:
            root.right = self.removeNode(root.right, val)
        
        
        # 上面是遍历过程， 下面是实际删节点的地方（分三种情况：1 没有child，2 1个child，3 2个child）
        else:                                             # val = root.left/right的时候      待删除节点等于 root.val 左右节点都存在的时候需要考虑拆分树
            if root.left == None and root.right == None:
                return None
            elif root.left == None:
                return root.right
            elif root.right == None:
                return root.left
            else:
                replacement = root.right                  # 这里是一直找到最右边？？
                while replacement.left is not None:
                    replacement = replacement.left
                root.val = replacement.val
                root.right = self.removeNode(root.right, replacement.val)

        return root


    def remove(self, val):
        self.root = self.removeNode(self.root, val)


    def searchNode(self, root, val):                # 递归函数一定要有一个递归出口
        if root is None or root.val == val:
            return root
        
        if val < root.val:
            return self.searchNode(root.left, val)
        else:
            return self.searchNode(root.right, val)
        

    def search(self, val):
        return self.searchNode(self.root, val)
    

    def inorder(self, node):
        if node is not None:
            self.inorder(node.left)
            print(node.val, end=" ")
            self.inorder(node.right)

    def inorderTraversal(self):
        self.inorder(self.root)
        print('')

def Test():
    bst = BinarySearchTree()
    bst.insert(5)
    bst.insert(3)
    bst.insert(7)
    bst.insert(2)
    bst.insert(4)
    bst.insert(6)
    bst.insert(8)

    print("Inorder Traversal:")
    bst.inorderTraversal()  # 输出: 2 3 4 5 6 7 8

    print("Searching for 4:", bst.search(4) is not None)  # 输出: True
    print("Searching for 10:", bst.search(10) is not None)  # 输出: False

    bst.remove(3)
    print("Inorder Traversal after removing 3:")
    bst.inorderTraversal()  # 输出: 2 4 5 6 7 8
Test()
```

```python
# 将有序数组转二叉搜索树

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def sortedArray(nums, l, r ):           # 以中间节点为root，左边做左子树，右边做右子树
    if l > r:
        return None
    
    mid = (l + r) // 2
    root = TreeNode(nums[mid])
    root.left = sortedArray(nums, l, mid - 1)
    root.right = sortedArray(nums, mid + 1, r)
    return root
    


def sortedArrayToBST(nums):
    return sortedArray(nums, 0, len(nums) - 1)





# 验证二叉搜索树  ---->进行中序遍历看看是否是递增的

#     2
#    / \
#   1   3

# 以这个树为例, 先调用inorder(1, list) 此时处理2的函数暂停
# 节点1的左边是None, 所以返回, 然后把1加入list, 然后处理节点1的右边, 也是None, 返回
# 此时处理2的函数恢复, 把2加入list, 然后处理节点2的右边, 调用inorder(3, list), 节点3的
# 左边是None, 所以返回, 然后把3加入list, 然后处理节点3的右边, 也是None, 返回

def inorder(node, inorder_list):
    if node is not None:
        inorder(node.left, inorder_list)        # 先一直做走到底,然后加根节点,然后走右子树
        inorder_list.append(node.val)
        inorder(node.right, inorder_list)


def isValidBST(root):
    inorder_list = []
    inorder(root, inorder_list)
    for i in range(1, len(inorder_list)):                   # 判断是否是递增的
        if inorder_list[i] <= inorder_list[i - 1]:
            return False
    return True

print(isValidBST(TreeNode(2, TreeNode(1), TreeNode(3))))  # 输出 True



# 合法二叉搜索树，中序遍历



def checkValidBST(root, min, max):       # 这里min和max是开区间
    if root is None:
        return True
    if root.val <= min or root.val >= max:              # 左子树一定要在min和val之间
        return False
    return checkValidBST(root.left, min, root.val) and checkValidBST(root.right, root.val, max)

def isValidBST(root):
    return checkValidBST(root, 0.001 , 100000)


print(isValidBST(TreeNode(2, TreeNode(1), TreeNode(3))))  # 输出 True
```
