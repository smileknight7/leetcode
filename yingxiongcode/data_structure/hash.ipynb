```python


# 好数对的数目


def numIdenticalPairs(nums):    #  O(n)2时间复杂度
    n = len(nums)
    cnt = 0
    for i in range(n):
        x = nums[i]
        for j in range(i):
            if nums[j] == x:   
                cnt += 1
    print(cnt)

# 哈希表做法

def numIdenticalPairs2(nums):   # O(n)时间复杂度
    n = len(nums)
    h = {}                          # 实际上转化成了统计计数问题
    cnt = 0
    for i in range(n):
        x = nums[i]
        cnt += h.get(x, 0)   # 统计当前数字之前出现的次数, d
        h[x] = h.get(x, 0) + 1
    print (cnt)
    
numIdenticalPairs([1, 2, 3, 1, 1, 3])
numIdenticalPairs2([1, 2, 3, 1, 1, 3])








# 这个数组有n+1个不同元素，同时这个数组有n个重复元素，找出这个重复N次的元素
# 在长度为2N的数组中找出重复N次的元素

def repeatedNTimes(nums):
    hash = {}
    n = len(nums) //2
    for x in nums:
        hash[x] = hash.get(x, 0) + 1              # 计数
    for k, v in hash.items():
        if v == n:
            return k
print(repeatedNTimes([1, 2, 3, 3]))


# 给一个整数数组arr， 统计数组中每个数的出现次数，如果每个数的出现次数是独一无二的就返回true，否则返回false
# 独一无二的出现次数

def uniqueOccurrences(arr):
    count = {}
    for x in arr:
        count[x] = count.get(x, 0) + 1
    hash = {}
    for v in count.values():
        if hash.get(v):                     # 之前如果出现过就返回false，否则返回true
            return False
        hash[v] = 1
    return True

print(uniqueOccurrences([1, 2, 2, 1, 1, 3]))
```
