```python


# *********
# o****o***


# 小明反转硬币



s = '*********'
t = 'o****o***'

n = len(s)
ret = 0
s = [i for i in s]
for i in range(n):
    if s[i] != t[i]:
        s[i] = ('*' if s[i] == 'o' else 'o')
        s[i + 1] = ('*' if s[i + 1] == 'o' else 'o')
        ret += 1
print(ret)


# 给定一个长度为n的整数数列A， A中i个元素为Ai的， 整数res初始为，如果一个数a存在于A且a+1, a+2均存在于A中，则res+1, 输出res值

A = [1, 2, 3, 5, 6, 7, 8]
res = 0
A = sorted(set(A))
for i in range(len(A)):
    if A[i] + 1 in A and A[i] + 2 in A:
        res += 1
print(res)


# 给一串长度为n， 且只含有小写英文字母的字符串S 用小括号()将字符串S分开， 使得每对小括号内仅有一个元音字母

def isYuanYiN(s):
    return s in ['a', 'e', 'i', 'o', 'u']


s = 'abecidofug'
n = len(s)
cnt = 0 
ret = '('

flag = False
for i in range(n):
    if isYuanYiN(s[i]):
        flag = True                                 # 确定原因至少出现过一次
        cnt += 1
        if cnt > 1:
            ret += ')('                             # 元音用于分割
            cnt = 1
    ret += s[i]                                     # 无论是否是元音都要加到ret中
ret += ')'
if flag == False:
    ret = s
print(ret)
```
