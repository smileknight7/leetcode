```python
# 双指针回纹判定


# 判断指针对应的元素是否相等，判定终点是i>= j
# abcba
# ^  ^
# i  j

s = 'abcba'
i = 0
j = len(s) - 1
while i < j:
    if s[i] != s[j]:
        break
    i += 1
    j -= 1

if i >= j:
    print('是回文')                 # 没有break的时候最终i>=j，说明是回文

```

```python
# 反转字符数组



# 用指针进行交换
s = 'ArithmeticError'
s = list(s) 
i = 0
j = len(s) - 1
while i < j:                 # 偶数的时候i会大于j，奇数的时候i会等于j（跳出循环的时候）
    s[i], s[j] = s[j], s[i]  # string是不可变类型，不能直接交换，需要转换为list
    i += 1
    j -= 1
s = ''.join(s)  # list转换为string
print(s)


# 等腰三角形
# 1, 2, 3, 4
# 2, 4, 6, 8
A = [1, 2, 3, 4]
B = [3, 4, 5, 6]
C = [2*x for x in A]

i = 0
j = 0
answer = 0
while i < len(A) and j < len(B):
    if C[i] > B[j]:
        answer += 1
        j += 1
        i += 1
    else:   
        i += 1
print(answer)




## 滑动窗口   初始化两个游标i 和 j， 其中i=0, j=-1, 代表开始是一个空窗口

# 朴素算法中j需要归位到i的位置，但是滑动窗口不需要
# 滑动窗口的复杂度O(n)  朴素算法的复杂度O(n^2)

# 挑选子字符串
# 有n个数，和一个整数m， 现在要从这n个数中挑选出来一个连续的字符串，要求这个字符串中有k个数大于等于m

a = [1, 2, 3, 4, 5, 6, 7, 8]
m = 9
k = 3
n = len(a)



def slidewindow(n, k, a):
    i , j = 0, -1
    sum = 0 
    answer = 0
    while j < n-1:
        j += 1
        sum += a[j]
        while sum > k:              # 这个代表i，j是满足条件的区间
            answer += n - j
            i += 1
            sum -= a[i]
    return answer

    
ans = slidewindow(n, k, a)
print(ans)




# 最长子字符串


def slidewindow(n, k, a):
    i , j = 0, -1
    count = {}
    ret = 0
    while j < n-1:
        j += 1
        count[a[j]] = count.get(a[j], 0) + 1
        while count[a[j]]  > k:             
            count[a[i]] -= 1
            i += 1

        ret = max(ret, j - i + 1)
        
    return ret

a = [1, 2, 3, 4, 5, 6, 7, 8]
k = 9
n = len(a)
ans = slidewindow(n, k, a)
print(ans)







#  给一个序列要求返回包含所有当前序列不同元素的子序列----其中最小的子序列是蓝桥序列，要求返回这个蓝桥序列大小

def sidewindow(n, a):
    i = 0
    j =-1
    count = {}
    total = 0  # 多少种不同
    ret = n    # 蓝桥序列长度最小值
    need = 0
    for x in a:
        if count.get(x, None) == None:           # 这里主要是记录总的种类数
            need += 1
        count[x] = 0
    total = 0
    ret = n

    while j < n-1:
        j += 1
        count[a[j]] += 1
        if count[a[j]] == 1:            # 出现一个新的字符
            total += 1
        while total == need:            # 合法蓝条序列
            ret = min(ret, j - i + 1)
            count[a[i]] -= 1
            if count[a[i]] == 0:
                total -= 1
            i += 1



n = 8
a = [1, 2, 3, 4, 5, 6, 7, 8]
answer = sidewindow(n, a)





```
