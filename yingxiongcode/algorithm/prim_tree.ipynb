# 最小生成树


## 贪心算法
## 适合非负边权和无向稠密图
## 顶点数最好小于1000


```python
# 建图
# 辅助数组定义源点为0，dist[i]表示顶点i与已生成的最小生成树之间的最小距离， visited[i]表示dist[i]是否已经确定，即i所在的最小生成树的edge是否确定
# 初始化， 所有数据如下 dist[i] = graph[0][i]，
#                     visited[i] = False，
#                     visited[0] = True
#                     sum = 0
# 找边权最小的点， 将dist[i]最小的点进行累加， visited[i] = True
# 统计edge权和





# 最小生成树要求的目标就是一个连通图，加上n-1条边， 所以如果有点没有被访问过，说明图不连通，返回-1     ！！！！！！！！

inf = 1e10


def initEdge(n):

    graph = [[inf]*n for _ in range(n)]             # 构建矩阵
    for i in range(n):
        graph[i][i] = 0
    return graph

def addEdge(graph, u, v, w):
    if w < graph[u][v]:
        graph[u][v] = w
    if w < graph[v][u]:
        graph[v][u] = w

def prim(n, graph):
    visited = [False] * n
    dist = [graph[0][i] for i in range(n)]              # dist初始化是初始点到别人的距离, 其实就是矩阵第一行
    sum = 0
    visited [0] = True
    
    while True:
        minDist = inf                                           # 因为有visited所以可以一直循环下去
        minIndex = -1
        for i in range(n):                                      # 0---n-1 中找一个最小的， 得到最小距离，和最最小点的index
                                                                # 假设有4个节点，前两行dist是这样的
            if not visited[i] and dist[i] < minDist:            # [0, 1, 4， 3]    [1, 0, 3, 4]
                minDist = dist[i]                               # 这里找到最小的距离是1号节点, 然后再那0号节点出发的距离去比比较并且更新 # [0, 1, 3, 3] 
                minIndex = i

        if minIndex == -1:
            break
        sum += minDist
        visited[minIndex] = True                                # 拿刚刚找到点，更新所有所有未进行访问的点的距离
        for i in range(n):
            if not visited[i] and graph[minIndex][i] < dist[i]:     
                dist[i] = graph[minIndex][i]                    # 更新邻居的距离

    for i in range(n):                      # 图不连通的时候的时候返回false
        if not visited[i]:
            return -1
    return sum
        

graph = initEdge(2021)
for i in range(1, 2022):                 # 城邦编号 1～2021
    for j in range(i + 1, 2022):
        u, v, w = i, j, 0
        while u or v:                    # 逐位比较，位数不够的当成 0
            if u % 10 != v % 10:         # 这一位不同，才把两个数字加进费用
                w += u % 10 + v % 10
            u //= 10
            v //= 10
        addEdge(graph, i - 1, j - 1, w)  # 编号减 1 当下标
print(prim(2021, graph))





# 城市最小通电费用
import math
def minCost(n, connections):
    graph = initEdge(n)
    x, y, h = [], [], []
    for a, b, c in connections:
        x.append(a)
        y.append(b)
        h.append(c)
    for i in range(n):
        for j in range(i+1, n):
            w =  math.sqrt((x[i]-x[j])**2 + (y[i]-y[j])**2) + (h[i]-h[j])**2
            addEdge(graph, i, j, w)
    ans = prim(n, graph)
    return ans
print(minCost(5, [[0,0,0],[1,1,1],[2,2,2],[3,3,3],[4,4,4]]))

# 繁忙的都市, 最大的最小值，和最小的最大值=-------> 一定是二分枚举








u, v, c = [1,1,2,2,3], [2,4,4,3,4], [3,5,7,6,8]
n, m = 4, 5




def check(max):
    graph = initEdge(n)
    for i in range(m):
        if c[i] <= max:
            addEdge(graph, u[i]-1, v[i]-1, c[i])
    ans = prim(n, graph)
    return ans != -1


l , r = 1, 100000

while l < r:
    mid = (l + r) // 2
    if check(mid):
        r = mid -1
        ans = mid
    else:
        l = mid + 1
print(ans)




```
