```python
# 一维数组的动态和

# ret[i] = ret[i -1] + nums[i]

def runningSum(nums):
    ret = [nums[0]]             # 第0个元素
    n = len(nums)
    for i in range(1, n):
        x = ret[i - 1] + nums[i]                # 这里是一个递推过程
        ret.append(x)
    return ret



# 找到数组中间位置

# def findMiddleIndex(nums):
#     n = len(nums)
#     for middleIndex in range(n):
#         if sum(nums[:middleIndex]) == sum(nums[middleIndex + 1:]):
#             return middleIndex
#     return -1

# 优化时间复杂度 ----> O(n2)  ---> O(n)   前缀和 +差分

def findMiddleIndex(nums):
    n = len(nums)
    s = [0 for i in range(n)]
    for i in range(1, n):                                   # 递推公式（因为不涉及到循环套循环所以优化到O(n)）
        s[i] = s[i - 1] + nums[i - 1]
    for middleIndex in range(n):
        a = 0
        if middleIndex == 0:
            a = 0
        else:
            a = s[middleIndex - 1]
        b = s[n - 1] - s[middleIndex]
        if a == b:
            return middleIndex
    return -1


nums = [1, 7, 3, 6, 5, 6]
print(findMiddleIndex(nums))




# 寻找数组中心下标









```
