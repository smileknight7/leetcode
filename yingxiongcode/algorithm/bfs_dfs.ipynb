# dfs 深度优先搜索

## 穷尽所有情况， 全排列，顶点总数小于11, 时间复杂度是On的阶乘


```python
# 全排列在计算过程中需要利用一个哈希表visited和一个stacked来记录当前访问的顶点
# 哈希表是为了快速查找某个顶点是否被访问，栈是为了记录顶点访问顺序
# 顶点访问，依次访问所有顶点，如果发现这个顶点不在hash中就将其加入到hash中，并将其压入栈中，然后继续访问下一个顶点
# 递归调用，利用递归调用自身继续访问下个顶点
# 递归出口，当本次访问的顶点等于节点总数的时候，栈中的元素就代表一个排列
# 回溯，回溯就是出栈过程，同时从哈希表中移除



maxn = 10
visited = [False for i in range(maxn)]

s = ''
sorted_str = ''
stack = [0 for i in range(maxn)]
ans = -1
count = 0


def dfs(depth, maxDepth):

    global count, ans, stack, visited
    if depth == maxDepth:                   # 递归出口
        # stacked中存储的就是[0, maxDepth-1]的一个排列
        find = True                         # 检查当前全排列和输入字符串是否匹配
        for i in range(maxDepth):
            idx = stack[i]
            if sorted_str[idx] != s[i]:
                find = False
                break

        if find:
            ans = count
        count += 1
        #print(stack, visited, count)        # debug部分
        return

    for i in range(maxDepth):               # 当前这一层的所有分支  
        if not visited[i]:
            visited[i] = True               # visited = [True, False, False]  [True, True, False]# 访问过了
            stack[depth] = i                # stack = [0, 0, 0]  [0, 1, 0]  这里的depath是上一层传进来的！！！！
            dfs(depth + 1, maxDepth)
            visited[i] = False              # 回溯 这个是倒着回溯的 [True, False, True]  [True, True, True]

# 这里的stack每次打印的就是一条路径

print(visited)

s = 'bca'
sorted_str = ''.join(sorted(s))               
dfs(0, len(s))                                
print('ans =', ans)                           




```

# 广度优先搜索

## 了解对列的概念
## 解决最短路径的问题
## 求解连通性问题
## 时间复度相对稳定

```python
# 构建hash表， visite 和对列queue记录当前访问的顶
# 起点插入对列， 添加到hash中
# 对列不为空时间， 弹出队首元素u，继续访问和u相邻的顶点v，如果v不在hash中则在hash中将v标记调度并且将来v插入对列



# 连通性问题




import queue


def initEdges(n):
    edges = [[] for _ in range(n)]
    return edges



def addEdge(edges, u, v):
    edges[u].append(v)


def bfs(edges, n, s):
    visited = [False for _ in range(n)]
    visited[s] = True
    q = queue.Queue()
    q.put(s)                    # 入队起点

    while not q.empty():            
        u = q.get()                      # 出队,队首元素
        for v in edges[u]:           # 遍历和u相邻的顶点v
            if not visited[v]:        # 如果v不在hash中
                visited[v] = True     # 在hash中将v标记调度
                q.put(v)              # 将来v插入对列
    return visited

def validPath(edges, n, s, t):
    init_edges = initEdges(n)
    for e in edges:
        addEdge(init_edges, e[0], e[1])             # 每条边代表第一个元素表示第几条边，第二个元素表示弧头/弧尾
        addEdge(init_edges, e[1], e[0])
    
    visited = bfs(init_edges, n, s)
    return visited[t]
```
