# 二分查找



```python

 
 
 
 
 
 
 
 
# 搜索插入位置
# 寻找数组中大于等于target最小值
def isGreen(nums, mid, target):
    return nums[mid] >= target
    
def bSearch(nums, target):
    l , r = -1, len(nums)                       # 确定边界
    while l + 1 < r: 
        mid = (l + r) // 2
        if isGreen(nums, mid, target):
            r = mid
        else:
            l = mid
    return r

def search_insert(nums, target):
    return bSearch(nums, target)
    


print(search_insert([1, 3, 5, 6], 4))



# 给定一个n个元素的有序数组，和一个目标值，写第一个函数搜索nums中的target，如果目标值存在就返回下标，否则返回-1


def search(nums, target):
    idx = bSearch(nums, target)
    if idx == len(nums) or nums[idx] != target:         # idx == len(nums) 大于数组中所有数，nums[idx] != target index对应的是一个大于target的值
        return -1
    return idx


print(search([1, 2, 3, 4, 5], 3))



# 在排序数组中查找元素的第一个和最后一个位置


def search_range(nums, target):
    idx = bSearch(nums, target)
    if idx == len(nums) or nums[idx] != target:         # 第二种情况是target在中间，原数组列表中不包含target这个元素
        return [-1, -1]
    ret = [idx, -1]
    ret[1] = bSearch(nums, target + 1) - 1   # 这里这样做的目标是target有重复元素的话就不呢个用idx+1 来确确保边界了
    return ret
print(search_range([1, 2, 3, 4, 5], 3))

    
```
