# 哈希算法，朴素算法时间度较高，用哈希可以

```python



# 给一个只包含大写字母的字符串，统计其中出现次数最多的字符，如果有多个字符出现次数相同，则返回字典序最小的那个字符。


cnt = {}
S ='DADADAD'
for c in S:
    cnt[c] = cnt.get(c, 0) + 1

maxc = 0        # 最大出现次数
ret = []        # 出现次数等于最大出现次数的字符列表
for v, c in cnt.items():                                # v是字符，c是出现次数
    if c > maxc:
        maxc = c
        ret = [v]
    elif c == maxc:
        ret.append(v)

ret = sorted(ret)    # 按字典序排序
print(ret)





# 给定字符串，请问其中有多少个不同的字符出现过？
s_list = ['asc', 'dad', 'dad', 'asf']
cnt2 = {}
for i in s_list:
    cnt2[i] = 1
print(len(cnt2))



# 优质数对
# 第一个元素表示，A,B 数组数量，第二个元素表示，A,B 数组中元素的范围

n = [3, [1, 2, 3], [3, 2, 1]]


# A[i] = B[j]          A[i] = B[j] 
# A[j] = B[i]   ---->  B[i] = A[j] 

# 要将两个数压缩成一个数字（状态压缩）
# (A[i] * SB + B[i]) =  (B[j] * SB + A[j])  

A = list(map(int, n[1]))
B = list(map(int, n[2]))


SB = 100001
cnt3 = {}
ret3 = 0
for j in range(3):
    x =  (B[j] * SB + A[j])
    ret3 += cnt3.get(x, 0)                  # 查询

    y = (A[j] * SB + B[j])                  # 插入
    cnt3 [y] = cnt3.get(y, 0) + 1
print(ret3)
```
