# 例题 二分查找




```python

def Findmintargetindex(array, target):
    l , r = -1, len(array)
    while l + 1 < r:
        mid = (l + r) // 2
        if array[mid] >= target:         # 如果处左边界就将下标换成mid，右边界同理  这里是这个二分的代码
            r = mid                     # 二分过程就是在不断向边界迭代的过程
        else:
            l = mid
    return r, array[r]

print(Findmintargetindex([1, 2, 3, 4, 5], 3))
```

```python
# 线性枚举

def GetMax(n, a):
    max = 0
    for i in range(n):
        if a[i] > max:
            max = a[i]
    return max
print(GetMax(5, [1, 2, 3, 4, 5]))


# 线性枚举的，时间复杂度是O(nm)
# 给一个数组，判断的是否存在连续三个数都是奇数
def threeConsecutiveOdds(arr):
    l = len(arr)
    for i in range(l - 2):
        if arr[i] % 2 == 1 and arr[i + 1] % 2 == 1 and arr[i + 2] % 2 == 1:
            return True
        
    return False
print(threeConsecutiveOdds([2, 6, 4, 1]))


# 计算最大连续1的个数(给定一个二进制数据，检查最大的连续1个数)
def findMaxConsecutiveOnes(nums):
    max = 0                         # 最大连续1个数
    pre = 0                         # 当前伦次的连续1
    for x in nums:
        if x == 1:
            pre += 1
            max = max if max > pre else pre
        else:
            pre = 0
    return max          

print(findMaxConsecutiveOnes([1, 1, 0, 1, 1, 1]))



# 有序数组中的单一元素(由于整数组成有序数组，其中每个元素都会出现两次，只有一个元素出现一次，返回出现的那个元素值)
def singleNonDuplicate(nums):
    l =len(nums)
    if l == 1:
        return nums[0]
    for i in range(1, l-1):
        if nums[i] != nums[i-1] and nums[i] != nums[i+1]:     # 问题转化成当前整个数不等于两边的数
            return nums[i]
    if nums[0] != nums[1]:        # 如果第一个数不等于第二个数，说明第一个数就是单一元素
        return nums[0]
    return nums[l-1]        
print(singleNonDuplicate([1, 1, 2, 3, 3, 4, 4, 8, 8]))



# 交换数字
def swapNumbers(numbers):
    numbers[0], numbers[1] = numbers[1], numbers[0]
    return numbers  
print(swapNumbers([1, 2]))



# 位1的个数
# 输入是一个二进制的整数，返回二进制中1 的个数
def hammingWeight(n):
    sum =0
    while n !=0:                    # 这里是在构建二进制？
        sum += n %2
        n = n // 2
    return sum
print('hammingWeight')
print(hammingWeight(11))  # 1011  输出3



# 返回数组中的一个index让index左边和右边的数组和相等
def finMiddleIndex(nums):
    l = len(nums)
    for i in range(l):
        if sum(nums[:i]) == sum(nums[i+1:]):   # 这里是判断左边的和是否等于右边的和
            return i
    return -1

print(finMiddleIndex([1, 7, 3, 6, 5, 6]))                   # 这个时间复杂度比较高还是O(n2)的时间复杂度
```
