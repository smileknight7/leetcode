```python
# 选择排序

# 在未排序的序列中找到最小的，和序列起始的数字进行交换， 
'''
时间复杂度较高，但是空间复杂度较小，稳定性差！！
'''

# 给定一个包含红色，白色，蓝色共n个元素的数组，原地进行排序，使得相同元素相邻，并且按照红色，白色，蓝色顺序排列。


def SortColors(nums):
    n = len(nums)
    for i in range(0, n):  
        minIndex = i                     # 保存最小值下标(注意，每次都是在剩下的内容中排序， 所以外层循环是从i遍历)
        for j in range(i + 1, n):
            if nums[j] < nums[minIndex]:
                minIndex = j
        nums[i], nums[minIndex] = nums[minIndex], nums[i]  # 交换位置
    return(nums)
print(SortColors([2,0,2,1,1,0]))


# 寻找两个正序数组的中位数
# 给定两个大小分别为m， n的正序数组，找出并返回这两个正序数组的中位数


def findMedianSortedArrays(nums1, nums2):
    for x in nums2:
        nums1.append(x)
    n = len(nums1)
    SortColors(nums1)
    if n % 2 == 1:
        return nums1[n // 2]
    else:
        return (nums1[n // 2 - 1] + nums1[n // 2]) / 2
print(findMedianSortedArrays([1, 3], [2]))



# 至少是其他数字两倍的最大数
# 给一数组nums， 其中总是存在唯一一个最大整数， 找出最大元素，并且检查它是否至少是数组中其他数字的两倍，如果是则返回最大元素的索引，否则返回-1

def dominantIndex(nums):
    max = 0
    for i in range(1, len(nums)):
        if nums[i] > nums[max]:
            max = i
    nums = SortColors(nums)
    maxv = nums[-1]
    maxsubv = nums[-2]
    if maxv >= 2 * maxsubv:
        return max
    return -1
    
print(dominantIndex([3, 6, 1, 0]))

```

```python
# 冒泡排序

# 比较相邻数字，大的数字放到右边
# 有两个循环，外层循环n-1， 内层是n-1, n-2, n-3 这种等差数列求和，最终整体是n2的时间复杂度
# 空间复杂度是O
# 稳定性好，适合小数据量排序




# 合并两个有序数组，两个按照非递减数据排列的整数数组nums1， num2, 合并nums1和nums2，使得num1成为一个非递减数组


def bubbleSort(nums):
    n = len(nums)
    for i in range(n-1, 0, -1):     # 逆序枚举   n-1 ---> 1
        for j in range(i):               # 正序       0 ---> i-1
            if nums[j] > nums[j+1]:
                nums[j], nums[j+1] = nums[j+1], nums[j]         # 交换位置

    

def merge(nums1, nums2):
    for i in range(len(nums2)):
        nums1.append(nums2[i])
    bubbleSort(nums1)
    print(nums1)

merge([1, 2, 3], [2, 5, 6])


# 给一个整数数组nums统计并且返回在nums中至少有1个严格较小元素和一个严格较大元素的元素数目
def countElements(nums):
    bubbleSort(nums)
    count = 0
    for i in nums:
        if i > nums[0] and i < nums[-1]:
            count += 1
    return count
print(countElements([11, 7, 2, 15]))



# 最后一块石头的重量
# 有一堆石头，每块石头的重量都是正整数， 每一回合选出两个最重的石头，然后将它们一起粉碎，假设石头的重量分别为x和y，且x <= y，那么粉碎的结果如下：
# 如果x == y，那么两块石头都会被完全粉碎；
# 如果x != y，那么重量为x的石头将会完全粉碎，而重量为y的石头新重量为y-x。
# 最后，最多只会剩下一 块石头。返回此石头的重量。如果没有石头剩下，返回0。

def lastStoneWeight(stones):
    n = len(stones)
    while n > 1:
        bubbleSort(stones)
        if stones[-1] == stones[-2]:
            stones.pop()
            stones.pop()
        else:
            stones[-2] = stones[-1] - stones[-2]
            stones.pop()
        n = len(stones)
    if n == 1:
        return stones[0]
    return 0
print(lastStoneWeight([2,7,4,1,8,1]))
```

```python
# 插入排序
# 将后面的元和前面已经排好序的序列进行比较，然后插入到排好序的序列中



# 计算平均工资
# 给一个salary数组，去掉最低工资和最高工资后，计算剩余工资的平均值


def insertionSort(nums):
    n = len(nums)
    for i in range(1, n):
        x = nums[i]
        j = i - 1
        while j >= 0 :
            if x < nums[j]:
                nums[j + 1] = nums[j]
            else:
                break
            j -= 1                  # 这里要一直向前排过去
        nums[j + 1] = x             # 插入到这个位置

def average(salary):
    salary.sort()
    return sum(salary[1:-1]) / len(salary[1:-1]) if len(salary) > 2 else 0      

print(average([4000,3000,1000,2000]))




# 给一个整数数组，删除最小5%的数字和最大5%的数字剩余数字的平均值

def trimMean(arr):
    insertionSort(arr)
    n = len(arr)
    print(n//20)
    print(n-n//20)
    print(len(arr[n//20:n-n//20]))
    return sum(arr[n//20:n-n//20]) / len(arr[n//20:n-n//20]) if n > 20 else 0

print(trimMean([6,2,7,5,1,2,0,3,10,2,5,0,5,5,6,8,7,6,8,0]))


# 学生分数的最小差值
# 给一个下标从0开始的整数数组，其中nums[i]表示第i名学生的分数，另给一个整数k，从数组中选出任意的k名学生，使k个分数间最高分和最低分数的差值达到最小

def minimumDifference(nums, k):
    insertionSort(nums)
    n = len(nums)
    ret = 10000000
    for i in range(0, n-k+1):               # 这个遍历的是区间数量
        l = i
        r = i + k - 1                       # r < n ---->  i + k -1 < n
        ret = min(ret, nums[r] - nums[l])
    return ret
print('minimumDifference')
print(minimumDifference([9,4,1,7], 2))
```

```python
# 计数排序， 基于哈希的排序方法，记录每个元素的出现次数，根据计数数组对原始数组进行排序
# 构建一个计数器， 将原数组，放到计数器中并且在计数器上计数， 将计数器依次枚举出来，实现排序
# 时间复杂度: 第一个循环 初始化计数器O(k), 枚举所有数字，哈希过程O(n), 最后枚举计数器O(k), 总体时间复杂度是O(n+k)
# 必须是整数， 限制范围！不要超过100000



def countingsort(a, m):
    count = [0 for i in range(m + 1)]  
    for x in a:
        count[x] += 1                           # count 的下标就代表实际的数字
    idx = 0
    for v in range(m + 1):
        while count[v] > 0:
            a[idx] = v
            idx += 1                            # idx是原数组下标用于替换原数组的值
            count[v] -= 1




# 最后一个石头的重量
def lastStone(nums):
    n = len(nums)
    while n > 1:
        countingsort(nums, max(nums))
        if nums[-1] == nums[-2]:
            nums.pop()
            nums.pop()
        else:
            nums[-2] = nums[-1] - nums[-2]
            nums.pop()
        n = len(nums)
    if n == 1:
        return nums[0]
    return 0
print(lastStone([2,7,4,1,8,1]))


# 学生成绩的区间范围
def mindifference(nums, k):
    countingsort(nums, max(nums))
    n = len(nums)
    ret = 10000000
    for i in range(0, n-k+1):               # 这个遍历的是区间数量
        l = i
        r = i + k - 1                       # r < n ---->  i + k -1 < n
        ret = min(ret, nums[r] - nums[l])
    return ret
```

```python
# 归并排序
# 一个非有序序列可以拆成两个非有序序列， 分别调用归并排序， 对两个有序序列等进行合并
# 合并的时候要使用双指针！！！
# 归并排序本质上是一个二叉树 ~， 时间复杂度O(nlogn)--->二叉树的高度, 空间复杂度O(n), 稳定性好，适合大数据量排序


def merge(left, right):
    ret = []
    i = j = 0
    n1 =  len(left)
    n2 = len(right)
    while i < n1 and j < n2:           # 这两个条件总有一个先不满足
        if left[i] <= right[j]:
            ret.append(left[i])
            i += 1
        else:
            ret.append(right[j])
            j += 1
    
    while i < n1:                       # 有剩余就将剩余的元素放到ret中
        ret.append(left[i])
        i += 1
    while j < n2:
        ret.append(right[j])
        j += 1
    return ret

def mergeSort(nums):

    if len(nums) <= 1:              #  递归先写出口
        return nums
    m = len(nums) // 2              # 中点
    l = mergeSort(nums[:m])              # 递归左边
    r = mergeSort(nums[m:])              # 递归右边
    return merge(l, r)                    # 合并两个有序数组

print(mergeSort([2,7,4,1,8,1]))



class ListNode:
    def __init__(self, val=0, next=None):         
        self.val = val
        self.next = next

# 给链表的头节点head， 将其升序排列， 返回排列之后的链表


class chainsort:

    def chainMergeSort(self, a, b):   # 有序链表通过这个归并  --> 得到升序排列链表
        a = self.sortList(a)
        b = self.sortList(b)
        head = ListNode()
        tail = head                             # tail代表链表尾节点，当前初始化为head
        
        while a or b:                           # 边界情况
            if a == None:
                tail.next = b
                break
            elif b == None:
                tail.next = a
                break
            
            elif a.val <= b.val:
                tail.next = a                   # tail节点变成a
                a = a.next                      # a变成a.next
            else:
                tail.next = b
                b = b.next
            tail = tail.next                    # tail一直代表链表的尾节点
            tail.next = None

        return head.next

    def sortList(self, head):           # 拆解链表
        if head  == None:
            return None
        elif head.next == None:
            return head
        
        fast = head                             # 定义指针初始化head
        slow = head
        prev = None
        while fast:
            prev = slow
            slow = slow.next
            fast = fast.next
            if fast:
                fast = fast.next                # 快指针走两步
        # head -> ... -> prev -> slow -> ... fast
        prev.next = None                        # 打断链表
        return self.chainMergeSort(head, slow)
```

```python

# 快速排序
# 选择一个基准元素，分割数组将比基准小的元素放到左边，大的元素放到右边
# 调用递归，最终实现排序

# 快速排序时间复杂度 是nlogn，最坏情况是n^2，空间复杂度是logn


# 存在重复元素，如果任意值在数组中出现至少两次
import random


# 1 3 7 6 4 5 2 
#         idx
# 4 3 7 6 1 5 2   交换
# 2 3 7 6 1 5 4
# 2 3 4 6 1 5 7   交换   
# 2 3 4 6 1 5 7   j-1  j-1
# 2 3 1 6 4 5 7   交换
# 2 3 1 4 6 5 7   交换且i= j
def Partition(nums, l, r):
    idx = random.randint(l, r)  
    nums[idx], nums[r] = nums[r], nums[idx]  
    i = l 
    j = r
    x = nums[i]
    while i < j:                                    # i和j向中间逼近
        while i < j and nums[j] > x:               #  满足条件i < j， 不满足 nums[j] > x 会跳出循环走if判断 
            j -= 1
        if i < j:                                   
            nums[i], nums[j] = nums[j], nums[i]
            i += 1
        while i < j and nums[i] < x:
            i += 1
        if i < j:
            nums[i], nums[j] = nums[j], nums[i]
            j -= 1
    return i                                    # 返回这个值之后i坐标的数字小于它， 右边的数字大于它

def QuickSort(nums, l, r):
    if l >= r:
        return
    pivot = Partition(nums, l, r)
    QuickSort(nums, l, pivot - 1)
    QuickSort(nums, pivot + 1, r)

def containsDuplicate(nums):
    n = len(nums)
    QuickSort(nums, 0, n - 1)
    for i in range(1, n):
        if nums[i] == nums[i - 1]:
            return True
    return False


print(containsDuplicate([1,3, 7, 6, 4, 5,2]))




# 给定一个大为n的nums，返回其中的多数元素，多数元素是指打数组中出现次数大于等于n/2的元素
def majorityElement(nums):
    n = len(nums)
    QuickSort(nums, 0, n - 1)
    return nums[n // 2] if n > 0 else None


majorityElement([3, 2, 3,1])
```

```python
# 桶排序
# 是一种基于计数的排序方法
# 桶排序是将数据分桶，然后桶内进行选择排序


# 给定一个字符串s， 根据字符串中出现的频率对其进行降序排序，一个字符出现的频率是它出现在字符串中的次数


# 关注数据原本的意思
# count的下标表示所有元素(128个ASCII码), 值表示这些元素的出现次数
# bucket下标表示计数次数，其中的一个对应元素就是一个ASCII码，对应的list

def bucketsort(nums, max):
    n = len(nums)
    bucket = [[] for i in range(n+1)]    # n +1 个桶
    count = [0 for i in range(max)]        # 计数器
    for i in range(n):
        count[ord(nums[i])] += 1
    for i in range(max):                 # cnt是对应的桶的次数
        cnt = count[i]
        bucket[cnt].append(chr(i))      # 出现相同次数的元素放到同一个桶中  ord是将字符变成整数，chr是将整数变成字符

    return bucket

def frequencySort(s):
    bucket = bucketsort(s, 128)          # 128个ASCII码
    ret = ''    
    for i in range(len(bucket)-1, -1, -1):   # 倒序遍历桶
        for j in range(len(bucket[i])):
            ret += bucket[i][j] * i         # 出现次数 * 字符
    return ret

print(frequencySort("tree"))
```

```python
# 基数排序，对所有元素进行收集和分配
# 获取所有元素的啊的最大位数，然后从个位开始，依次对每一位进行排序，直到最高位
# 每个桶是个对列

# 时间复杂度是O(d*(n+b))，d是最大位数，n是元素个数，b是桶的个数

MAXN = 50005
MAXT = 8                        # 最多8位数，8次桶排序
BASE = 10                       # 按10进制拆，

def RadixSort(nums):
    n = len(nums)
    PowOfBase = [1 for i in range(MAXT)]
    for i in range(MAXT):
        PowOfBase[i] = PowOfBase[i - 1] * BASE if i > 0 else 1


    for i in range(n):
        nums[i] += PowOfBase[MAXT -1]

    pos = 0 
    while pos < MAXT:                                   # 基数排序实际上是MAXT次的桶排序
        RadixBucket = [[] for i in range(BASE)]         # 这里就是实际进行的桶排序
        
        for i in range(n):
            rdx = nums[i] // PowOfBase[pos] % BASE
            RadixBucket[rdx].append(nums[i])

        top = 0 
        for i in range(BASE):
            for rb in RadixBucket[i]:
                nums[top] = rb
                top += 1
        pos += 1


    for i in range(n):
        nums[i] -= PowOfBase[MAXT -1]


def sortArray(nums):
    RadixSort(nums)
```
