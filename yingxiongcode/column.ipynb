```python

# 顺序表实现对列




class Queue:
    def __init__(self):
        self.data = []
        self.head = 0
        self.tail = 0
    
    def push(self, val):
        self.data.append(val)
        self.tail += 1
    
    def pop(self):
        if self.head == self.tail:
            raise IndexError("pop from empty queue")
        val = self.data[self.head]          # 队首指针
        self.head += 1             
        return val
    
    def front(self):
        if self.head == self.tail:
            raise IndexError("front from empty queue")
        return self.data[self.head]

    def size(self):
        return self.tail - self.head

    def empty(self):
        return self.size() == 0
    
def Test():
    q = Queue()
    q.push(1)
    q.push(2)
    q.push(3)
    assert q.size() == 3
    assert not q.empty()
    assert q.front() == 1
    assert q.pop() == 1
    assert q.front() == 2
    assert q.pop() == 2
    assert q.pop() == 3
    assert q.empty()
    print("All tests passed.")
Test()
    
```

```python
# 链表实现的对列


class Node:
    def __init__(self, val):
        self.val = val
        self.next = None                    # 初始化的时候是孤立节点所以指向是None

class Queue:
    def __init__(self):
        self.head = None
        self.tail = None
        self.len = 0
    
    def push(self, val):
        new_node = Node(val)
        if self.tail is None:               # 如果队列为空
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node        # 将新节点加入到队尾
            self.tail = new_node             # 更新队尾指针
        self.len += 1
    
    def pop(self):
        if self.empty():               # 校验是否是空对列
            raise IndexError("pop from empty queue")
        val = self.head.val                 # 队首指针
        self.head = self.head.next          # 更新队首指针
        if self.head is None:               # 如果队列为空，更新队尾指针
            self.tail = None
        self.len -= 1
        return val
    
    def front(self):
        if self.empty():
            raise IndexError("front from empty queue")
        return self.head.val

    def size(self):
        return self.len

    def empty(self):
        return self.len == 0
    
def Test():
    q = Queue()
    q.push(1)
    q.push(2)
    q.push(3)
    assert q.size() == 3
    assert not q.empty()
    assert q.front() == 1
    assert q.pop() == 1
    assert q.front() == 2
    assert q.pop() == 2
    assert q.pop() == 3
    assert q.empty()
    print("All tests passed.")
Test()
```

```python

# 最近请求次数


class RecentCounter:
    def __init__(self):
        self.queue = []
        self.front = 0
    
    def ping(self, t: int) -> int:
        self.queue.append(t)
        while self.queue[self.front] < t - 3000:                # t - self.queue[self.front] > 3000
            self.front += 1
        return len(self.queue) - self.front


recentcounter = RecentCounter()
print(recentcounter.ping(1))      # 返回 1
print(recentcounter.ping(100))    # 返回 2
print(recentcounter.ping(3001))   # 返回 3
print(recentcounter.ping(3002))   # 返回 3


# 数据流中的移动平均值

class MovingAverage:
    def __init__(self, size: int):
        self.queue = []
        self.front = 0
        self.size = size
        self.sum = 0.0                  # 对列中所有元素的和

    def get_size(self):
        return len(self.queue) - self.front
    
    def next(self, val: int) -> float:
        self.queue.append(val)
        self.sum += val 
        while self.get_size() > self.size:                  # 出队操作
            self.sum -= self.queue[self.front]
            self.front += 1
        return self.sum / self.get_size() if self.get_size() > 0 else 0.0       
    
movingaverage = MovingAverage(3)
print(movingaverage.next(1))  # 返回 1.0
print(movingaverage.next(10)) # 返回 5.5
print(movingaverage.next(3))  # 返回 4.66667
print(movingaverage.next(5))  # 返回 6.0
```

```python

# 用栈实现对列
# 对列的底层容器不是列表也不是链表，而是栈（再这种情况下），需要构建两个栈，s1，s2

# 栈是先进后出，但是对列是先进先出，多以要设计两个栈来实现这个过程，先将s1压入s2，然后再进行弹出




class Stack:
    def __init__(self):
        self.data = []
    def push(self, val):
        self.data.append(val)
    
    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self.data.pop()
    def top(self):
        if self.is_empty():
            raise IndexError("top from empty stack")
        return self.data[-1]
    def size(self):
        return len(self.data)
    def is_empty(self):
        return len(self.data) == 0


class MyQueue:
    def __init__(self):
        self.s1 = Stack()
        self.s2 = Stack()

    def push(self, x: int) -> None:
        self.s1.push(x)                     # 塞到主栈栈顶
        
    def pop(self) -> int:
        if self.s2.is_empty():                  # 如果辅助栈为空，则将主栈的元素全部压入辅助栈
            while not self.s1.is_empty():
                self.s2.push(self.s1.pop())
        return self.s2.pop()                    # 弹出辅助栈的栈顶元素

    def peek(self) -> int:
        if self.s2.is_empty():                  # 如果辅助栈为空，则将主栈的元素全部压入辅助栈
            while not self.s1.is_empty():
                self.s2.push(self.s1.pop())
        return self.s2.top()                    # 返回辅助栈的栈顶元素

    def empty(self) -> bool:
        return self.s1.is_empty() and self.s2.is_empty()
myqueue = MyQueue()






# 用对列实现栈
# 需要构建两个对列，主对列q1， 辅助对列q2
# 先将q1中元素依次出队到q2中，到最后一个元素的时候直接弹出 ---->然后往复运动

class Queue:
    def __init__(self):
        self.data = []
        self.head = 0
        self.tail = 0
    
    def push(self, val):
        self.data.append(val)
        self.tail += 1
    
    def pop(self):
        if self.head == self.tail:
            raise IndexError("pop from empty queue")
        val = self.data[self.head]          # 队首指针
        self.head += 1             
        return val
    
    def front(self):
        if self.head == self.tail:
            raise IndexError("front from empty queue")
        return self.data[self.head]

    def size(self):
        return self.tail - self.head

    def empty(self):
        return self.size() == 0


class Mystack:
    def __init__(self):
        self.q1 = Queue()
        self.q2 = Queue()
    
    def push(self, x: int) -> None:
        self.q1.push(x)                     # 将元素压入主对列

    def pop(self):
        while self.q1.size() > 1:              # 将主对列的元素依次出队到辅助对列，直到只剩下一个元素
            self.q2.push(self.q1.pop())
        val = self.q1.pop()                    # 弹出主对列的最后一个
        while self.q2.size() > 0:              # 将辅助对列的元素依次出队到主对列
            self.q1.push(self.q2.pop())
        return val
    def top(self):
        while self.q1.size() > 1:              # 将主对列的元素依次出队到辅助对列，直到只剩下一个元素
            self.q2.push(self.q1.pop())
        val = self.q1.pop()                    # 弹出主对列的最后一个
        self.q2.push(val)                      # 将弹出的元素压入辅助对列(这样久相当于是没有损失值)
        while self.q2.size() > 0:              # 将辅助对列的元素依次出队到主对列
            self.q1.push(self.q2.pop())
        return val

    def empty(self):
        return self.q1.empty()
```
