
# 栈，主要用于实现函数调用实现
# 栈   遵循后进先出的原则, 栈只能先访问顶部的盘子，修改和删除也只能针对顶部





```python


# 顺序表的栈实现

class Stack:
    def __init__(self):
        self.data = []
    def push(self, val):
        self.data.append(val)
    
    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self.data.pop()
    def top(self):
        if self.is_empty():
            raise IndexError("top from empty stack")
        return self.data[-1]
    def size(self):
        return len(self.data)
    def is_empty(self):
        return len(self.data) == 0

def Test():
    stk = Stack()
    stk.push(1)
    stk.push(2)
    stk.push(3)
    while not stk.is_empty():
        print(stk.pop())

Test()
```

```python

# 链表的栈实现


class Node:
    def __init__(self, val):
        self.val = val
        self.next = None



class Stack:
    def __init__(self):
        self.head = None                 # 空栈， 空列表
        self.len = 0
   
   
   
    def push(self, val):         # 入栈     链表头插法
        new_node = Node(val)
        new_node.next = self.head        # 将之前的栈头放到新节点后面
        self.head = new_node             # 将新节点放到栈的最前面
        self.len += 1       
    

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        val = self.head.val
        self.head = self.head.next         # 将表头后面的一个节点构建成链表头
        self.len -= 1
        return val



    def top(self):
        if self.is_empty():
            raise IndexError("top from empty stack")
        return self.head.val
    
    def size(self):
        return self.len
    def is_empty(self):
        return self.len == 0    



def Test1():
    stk = Stack()
    stk.push(1)
    stk.push(2)
    stk.push(3)
    while not stk.is_empty():
        print(stk.pop())
Test1()


```
