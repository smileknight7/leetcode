# 顺序数据结构在增删改查的时候要移动大量数据，链表在增删改查的时候只需要移动指针，所以链表的增删改查效率更高。


第一步：判断插入位置是否合法，
第二部：对给定元素生成一个链表终点
第三步：插入节点
第四步：更新链表

```python
class ListNode:
    def __init__(self, val=0, next=None):               # 数据域和指针域
        self.val = val
        self.next = next

class LinkedList:
    def __init__(self):
        self.head = None                                        # head表示的就是指针
        self.len = 0
    def size(self):
        return self.len
    def insert(self, pos, val):
        if pos < 0  or pos > self.size():
            raise IndexError("Index out of bounds")
        new_node = ListNode(val)
        if pos == 0:
            new_node.next = self.head           # 节点后继指向新的链表头
            self.head = new_node                # 修改链表头为新节点
        else:
            prev = self.head
            for _ in range(pos - 1):
                prev = prev.next
            new_node.next = prev.next           # 新节点的后继指向前一个节点
            prev.next = new_node                # 前一个节点的后继指向新节点
        self.len += 1
    def delete(self, pos):
        if pos < 0 or pos >= self.size():
            raise IndexError("Index out of bounds")
        if pos == 0:
            self.head = self.head.next          # 删除头节点，链表头指向下一个节点
        else:
            prev = self.head
            for _ in range(pos - 1):
                prev = prev.next
            prev.next = prev.next.next          # 前一个节点的后继指向下一个节点的后继
        self.len -= 1

    def update(self, pos, val):
        if pos < 0 or pos >= self.size():
            raise IndexError("Index out of bounds")
        current = self.head
        for _ in range(pos):
            current = current.next
        current.val = val                      # 更新节点的值

    def search(self, val):
        current = self.head
        while current:
            if current.val == val:                 # 返回找到的节点索引
                return current                     # 返回找到的节点
            current = current.next
        return None            

    def index(self, val):
        current = self.head
        index = 0
        while current:
            if current.val == val:
                return index                      # 返回找到的节点索引
            current = current.next
            index += 1
        return -1                                 # 未找到返回-1   
    def print(self):
        current = self.head
        while current:
            print(current.val, end=" -> ")
            current = current.next
        print("None")

    def build(self, vals):
        dummy = ListNode()
        tail = dummy
        for v in vals:
            tail.next = ListNode(v)
            tail = tail.next
        return dummy.next
    def show(self, head):
        current = head
        while current:
            print(current.val, end=" -> ")
            current = current.next
        print("None")
```
